```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Sesuaikan path ini dengan lokasi foto referensi siang hari dari Dihar
path = r"C:\Users\bangb\Project-TJ\data\reference\c04_gerbang_pemuda_20261023_094301.jpg"
img = cv2.imread(path)

# Pengecekan jika gambar tidak ditemukan
if img is None:
    print(f"Gambar tidak ditemukan di path: {path}")
else:
    # BGR ke RGB untuk matplotlib
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Proses CLAHE pada kanal L di ruang warna LAB
    lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)
    l, a, b = cv2.split(lab)
    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
    clahe_img = cv2.cvtColor(cv2.merge([clahe.apply(l), a, b]), cv2.COLOR_LAB2BGR)
    clahe_rgb = cv2.cvtColor(clahe_img, cv2.COLOR_BGR2RGB)

    # Menampilkan gambar sebelum dan sesudah CLAHE
    fig, ax = plt.subplots(1, 2, figsize=(12, 5))
    ax[0].imshow(img_rgb)
    ax[0].set_title('Sebelum (Asli)')
    ax[0].axis('off')

    ax[1].imshow(clahe_rgb)
    ax[1].set_title('Sesudah CLAHE')
    ax[1].axis('off')

    plt.show()
```

```python
# Terapkan Gamma Correction pada hasil CLAHE
gamma = 2.0 # Nilai dinaikkan agar lebih terang (bisa diubah antara 1.5 - 2.5)
lut = ((np.arange(256) / 255.0) ** (1 / gamma) * 255).astype('uint8')
gamma_img = cv2.LUT(clahe_img, lut)

# BGR ke RGB untuk matplotlib
gamma_rgb = cv2.cvtColor(gamma_img, cv2.COLOR_BGR2RGB)

# Menampilkan perbandingan
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
ax[0].imshow(clahe_rgb)
ax[0].set_title('Hasil CLAHE (Masih Gelap)')
ax[0].axis('off')

ax[1].imshow(gamma_rgb)
ax[1].set_title(f'Sesudah CLAHE + Gamma (γ={gamma})')
ax[1].axis('off')

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

# 1. Load Gambar Asli (Malam Hari)
path = '../data/reference/c04_gerbang_pemuda_20261023_181901_malam.jpg'
img_asli = cv2.imread(path)
img_asli_rgb = cv2.cvtColor(img_asli, cv2.COLOR_BGR2RGB)

# 2. Proses CLAHE
lab = cv2.cvtColor(img_asli, cv2.COLOR_BGR2LAB)
l, a, b = cv2.split(lab)
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
clahe_img = cv2.cvtColor(cv2.merge([clahe.apply(l), a, b]), cv2.COLOR_LAB2BGR)

# 3. Proses Gamma Correction
gamma = 2.5
lut = ((np.arange(256) / 255.0) ** (1 / gamma) * 255).astype('uint8')
img_final = cv2.LUT(clahe_img, lut)
img_final_rgb = cv2.cvtColor(img_final, cv2.COLOR_BGR2RGB)

# 4. Tampilkan Perbandingan BEFORE vs AFTER
fig, ax = plt.subplots(1, 2, figsize=(15, 6))

ax[0].imshow(img_asli_rgb)
ax[0].set_title('BEFORE (Gambar Asli Malam)')
ax[0].axis('off')

ax[1].imshow(img_final_rgb)
ax[1].set_title(f'AFTER (CLAHE + Gamma γ={gamma})')
ax[1].axis('off')

plt.tight_layout()
plt.show()
```
